```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict,Literal
```

```python
class QuadState(TypedDict):
    a:int
    b:int
    c:int
    
    equation:str
    discriminant:float
    result:str
```

```python
def showEQN(state:QuadState):
    equation=f"{state['a']}X2{state['b']}X{state['c']}"
    return {'equation':equation}
```

```python
def calcDisc(state:QuadState):
    disc=state['b']**2 - (4*state['a']*state['c'])
    return {'discriminant':disc}
```

```python
def realRoots(state:QuadState):
    root1=(-state['b']+state['discriminant']**0.5)/(2*state['a'])
    root2=(-state['b']-state['discriminant']**0.5)/(2*state['a'])
    res=f"The roots are {root1} and {root2}"
    return {'result':res}
```

```python
def repeatedRoots(state:QuadState):
    root=(-state['b'])/(2*state['a'])
    res=f"The only repeating root is {root}"
    return {'result':res}
```

```python
def noRealRoots(state:QuadState):
    res=f"No real roots!"
    return {'result':res}
```

```python
def checkkCondition(state:QuadState) -> Literal['realRoots','repeatedRoots','noRealRoots']:
    if state['discriminant']==0 : 
        return "repeatedRoots"
    elif state['discriminant']>0:
        return "realRoots"
    else:
        return "noRealRoots"
```

```python
graph=StateGraph(QuadState)

graph.add_node('showEQN',showEQN)
graph.add_node('calcDisc',calcDisc)

graph.add_node('realRoots',realRoots)
graph.add_node('repeatedRoots',repeatedRoots)
graph.add_node('noRealRoots ',noRealRoots)

graph.add_edge(START,'showEQN')
graph.add_edge('showEQN','calcDisc')
graph.add_conditional_edges('calcDisc',checkkCondition)

graph.add_edge('realRoots',END)
graph.add_edge('repeatedRoots',END)
graph.add_edge('noRealRoots',END)

workflow=graph.compile()
```

```python
initial_state={
    'a':4,'b':-5,'c':-4
}

workflow.invoke(initial_state)
```
